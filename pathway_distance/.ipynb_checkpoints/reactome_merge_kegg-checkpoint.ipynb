```python
import sqlite3
import pandas as pd
import os

# ============================================================================
# 0. CREATE DATABASE FROM SQL DUMP FILE (OPTIMIZED FOR LARGE FILES)
# ============================================================================

def create_database_from_dump_optimized(sql_dump_path, db_output_path='gk_current.db'):
    """
    Create a SQLite database from a SQL dump file, processing in chunks
    """
    print(f"Creating database from SQL dump: {sql_dump_path}")
    print("(Processing large file in chunks...)")
    
    conn = sqlite3.connect(db_output_path)
    cursor = conn.cursor()
    
    # Disable synchronous mode for faster writes
    cursor.execute("PRAGMA synchronous = OFF")
    cursor.execute("PRAGMA journal_mode = MEMORY")
    
    statement_count = 0
    error_count = 0
    
    try:
        with open(sql_dump_path, 'r', encoding='utf-8', errors='ignore') as f:
            statement = ""
            for line_num, line in enumerate(f):
                # Skip comments and empty lines
                if line.strip().startswith('--') or not line.strip():
                    continue
                
                statement += line
                
                # Execute when we hit a semicolon
                if statement.strip().endswith(';'):
                    try:
                        cursor.execute(statement)
                        statement_count += 1
                        
                        # Commit every 1000 statements
                        if statement_count % 1000 == 0:
                            conn.commit()
                            print(f"  Processed {statement_count} statements...")
                        
                        statement = ""
                    except Exception as e:
                        error_count += 1
                        if error_count <= 5:  # Only print first 5 errors
                            print(f"  Warning (line {line_num}): {str(e)[:80]}")
                        statement = ""
        
        conn.commit()
        print(f"✓ Database created successfully!")
        print(f"  Total statements executed: {statement_count}")
        print(f"  Errors encountered: {error_count}")
    
    except Exception as e:
        print(f"Error: {e}")
    
    finally:
        conn.close()
    
    return db_output_path


# ============================================================================
# 1. INSPECT DATABASE STRUCTURE (MINIMAL OUTPUT)
# ============================================================================

def inspect_database(db_path):
    """Inspect the database structure"""
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()
    
    # Get all table names
    cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
    tables = [t[0] for t in cursor.fetchall()]
    
    print(f"\nTables found: {tables}\n")
    
    # Show schema and row count for each table
    for table in tables:
        cursor.execute(f"PRAGMA table_info({table})")
        columns = [col[1] for col in cursor.fetchall()]
        
        cursor.execute(f"SELECT COUNT(*) FROM {table}")
        count = cursor.fetchone()[0]
        
        print(f"{table}:")
        print(f"  Columns: {columns}")
        print(f"  Rows: {count}\n")
    
    conn.close()
    return tables


# ============================================================================
# 2. EXTRACT INTERACTIONS (SMART DETECTION)
# ============================================================================

def extract_reactome_interactions(db_path, output_csv='reactome_interactions.csv'):
    """
    Extract unique protein interactions from Reactome database
    """
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()
    
    # Get all tables
    cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
    tables = [t[0] for t in cursor.fetchall()]
    
    print(f"Searching for interaction data in {len(tables)} tables...\n")
    
    df = None
    
    # Look for tables that likely contain interactions
    for table in tables:
        cursor.execute(f"PRAGMA table_info({table})")
        columns = [col[1].lower() for col in cursor.fetchall()]
        
        # Check if this table looks like it contains interactions
        has_protein_cols = any(
            'protein' in col or 'gene' in col or 'interactor' in col 
            for col in columns
        )
        has_pathway_col = any(
            'pathway' in col or 'reaction' in col 
            for col in columns
        )
        
        if has_protein_cols:
            print(f"Checking table: {table}")
            print(f"  Columns: {columns}")
            
            # Try to extract from this table
            try:
                # Get first few rows to understand structure
                sample = pd.read_sql_query(f"SELECT * FROM {table} LIMIT 5", conn)
                print(f"  Sample data:")
                print(sample.to_string())
                print()
                
                # If this looks promising, extract all unique pairs
                if len(sample) > 0:
                    # Find protein/gene columns
                    protein_cols = [c for c in sample.columns if 'protein' in c.lower() or 'gene' in c.lower()]
                    pathway_cols = [c for c in sample.columns if 'pathway' in c.lower() or 'reaction' in c.lower()]
                    
                    if len(protein_cols) >= 2:
                        print(f"  → Found protein columns: {protein_cols}")
                        if pathway_cols:
                            print(f"  → Found pathway column: {pathway_cols[0]}")
                        
                        # Build query
                        if pathway_cols:
                            query = f"""
                            SELECT DISTINCT
                                {protein_cols[0]},
                                {protein_cols[1]},
                                {pathway_cols[0]}
                            FROM {table}
                            WHERE {protein_cols[0]} IS NOT NULL 
                                AND {protein_cols[1]} IS NOT NULL
                            LIMIT 100000
                            """
                        else:
                            query = f"""
                            SELECT DISTINCT
                                {protein_cols[0]},
                                {protein_cols[1]}
                            FROM {table}
                            WHERE {protein_cols[0]} IS NOT NULL 
                                AND {protein_cols[1]} IS NOT NULL
                            LIMIT 100000
                            """
                        
                        df = pd.read_sql_query(query, conn)
                        print(f"  ✓ Extracted {len(df)} interactions from {table}\n")
                        break
            
            except Exception as e:
                print(f"  ✗ Error: {str(e)[:100]}\n")
    
    if df is not None and not df.empty:
        df.to_csv(output_csv, index=False)
        print(f"✓ Saved {len(df)} interactions to {output_csv}")
    else:
        print("✗ No interactions found")
    
    conn.close()
    return df


# ============================================================================
# 3. CLEAN AND DEDUPLICATE
# ============================================================================

def clean_interactions(df, output_csv='unique_interactions_cleaned.csv'):
    """Remove duplicates and clean interaction data"""
    if df is None or df.empty:
        print("No data to clean")
        return None
    
    print(f"\nCleaning {len(df)} interactions...")
    
    # Get column names
    cols = df.columns.tolist()
    protein1_col = cols[0]
    protein2_col = cols[1]
    
    # Remove self-interactions
    df = df[df[protein1_col] != df[protein2_col]].copy()
    print(f"  After removing self-interactions: {len(df)}")
    
    # Create normalized pair
    df['pair'] = df.apply(
        lambda row: tuple(sorted([str(row[protein1_col]), str(row[protein2_col])])),
        axis=1
    )
    
    # Remove duplicates
    df_unique = df.drop_duplicates(subset=['pair'], keep='first')
    df_unique = df_unique.drop('pair', axis=1)
    
    print(f"  After deduplication: {len(df_unique)}")
    
    df_unique.to_csv(output_csv, index=False)
    print(f"✓ Saved to {output_csv}")
    
    return df_unique


```

```python
# ============================================================================
# 5. MAIN EXECUTION
# ===========
sql_dump_path = './gk_current.sql'
db_path = './gk_current.db'

# Step 0: Create database
print("STEP 0: Creating database from SQL dump...")
print("="*60)
db_path = create_database_from_dump_optimized(sql_dump_path, db_path)

# Step 1: Inspect
print("\n\nSTEP 1: Inspecting database...")
print("="*60)
inspect_database(db_path)

# Step 2: Extract
print("\nSTEP 2: Extracting interactions...")
print("="*60)
df = extract_reactome_interactions(db_path)

# Step 3: Clean
if df is not None:
    print("\nSTEP 3: Cleaning interactions...")
    print("="*60)
    clean_interactions(df)
```

```python
import requests
import pandas as pd
import time
import os
import pickle

# ============================================================================
# 1. DOWNLOAD REACTOME INTERACTIONS FILE
# ============================================================================

def download_interactions_file(output_file='reactome_interactions_human.txt'):
    """
    Download human protein-protein interactions from Reactome
    """
    print("Downloading Reactome human protein-protein interactions...")
    
    url =  'https://download.reactome.org/96/interactors/reactome.homo_sapiens.interactions.tab-delimited.txt'

    
    try:
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        
        # Save to file
        with open(output_file, 'w') as f:
            f.write(response.text)
        
        # Parse into dataframe
        df = pd.read_csv(output_file, sep='\t')
        print(f"✓ Downloaded interactions file")
        print(f"  Shape: {df.shape}")
        print(f"  Columns: {df.columns.tolist()}\n")
        
        return df
    
    except Exception as e:
        print(f"✗ Error: {e}")
        return None


# ============================================================================
# 2. DOWNLOAD PATHWAY MAPPING FILES
# ============================================================================

def download_uniprot_to_pathways(output_file='uniprot_to_pathways.txt'):
    """
    Download UniProt to Pathways mapping (all levels)
    """
    print("Downloading UniProt to Pathways mapping...")
    
    url = 'https://download.reactome.org/96/UniProt2Reactome_All_Levels.txt'
    
    try:
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        
        with open(output_file, 'w') as f:
            f.write(response.text)
        
        df = pd.read_csv(output_file, sep='\t', header=None)
        df.columns = ['uniprot_id', 'reactome_id', 'url', 'pathway_name', 'evidence_code', 'species']
        
        print(f"✓ Downloaded UniProt to Pathways mapping")
        print(f"  Shape: {df.shape}")
        print(f"  Columns: {df.columns.tolist()}\n")
        
        return df
    
    except Exception as e:
        print(f"✗ Error: {e}")
        return None


def download_ensembl_to_pathways(output_file='ensembl_to_pathways.txt'):
    """
    Download ENSEMBL to Pathways mapping (all levels)
    """
    print("Downloading ENSEMBL to Pathways mapping...")
    
    url = 'https://download.reactome.org/96/Ensembl2Reactome_All_Levels.txt'
    
    try:
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        
        with open(output_file, 'w') as f:
            f.write(response.text)
        
        df = pd.read_csv(output_file, sep='\t', header=None)
        df.columns = ['ensembl_id', 'reactome_id', 'url', 'pathway_name', 'evidence_code', 'species']
        
        print(f"✓ Downloaded ENSEMBL to Pathways mapping")
        print(f"  Shape: {df.shape}")
        print(f"  Columns: {df.columns.tolist()}\n")
        
        return df
    
    except Exception as e:
        print(f"✗ Error: {e}")
        return None


def download_ncbi_to_pathways(output_file='ncbi_to_pathways.txt'):
    """
    Download NCBI Gene ID to Pathways mapping (all levels)
    """
    print("Downloading NCBI Gene ID to Pathways mapping...")
    
    url = 'https://download.reactome.org/96/NCBI2Reactome_All_Levels.txt'
    
    try:
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        
        with open(output_file, 'w') as f:
            f.write(response.text)
        
        df = pd.read_csv(output_file, sep='\t', header=None)
        df.columns = ['ncbi_gene_id', 'reactome_id', 'url', 'pathway_name', 'evidence_code', 'species']
        
        print(f"✓ Downloaded NCBI Gene ID to Pathways mapping")
        print(f"  Shape: {df.shape}")
        print(f"  Columns: {df.columns.tolist()}\n")
        
        return df
    
    except Exception as e:
        print(f"✗ Error: {e}")
        return None


# ============================================================================
# 3. DOWNLOAD PATHWAY INFORMATION
# ============================================================================

# Fix the pathway hierarchy download function
def download_pathway_hierarchy(output_file='pathway_hierarchy.txt'):
    """
    Download pathway hierarchy information
    """
    print("Downloading pathway hierarchy...")
    
    url = 'https://download.reactome.org/96/ReactomePathwaysRelation.txt'
    
    try:
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        
        with open(output_file, 'w') as f:
            f.write(response.text)
        
        # Read without assigning column names first
        df = pd.read_csv(output_file, sep='\t', header=None)
        
        # Assign correct number of columns based on actual data
        if df.shape[1] == 4:
            df.columns = ['parent_pathway_id', 'parent_pathway_name', 'child_pathway_id', 'child_pathway_name']
        elif df.shape[1] == 2:
            df.columns = ['parent_pathway_id', 'child_pathway_id']
        
        print(f"✓ Downloaded pathway hierarchy")
        print(f"  Shape: {df.shape}\n")
        
        return df
    
    except Exception as e:
        print(f"✗ Error: {e}")
        return None

print("Fixed function ready to use!")


def download_pathway_list(output_file='pathway_list.txt'):
    """
    Download complete list of pathways
    """
    print("Downloading complete pathway list...")
    
    url = 'https://download.reactome.org/96/ReactomePathways.txt'
    
    try:
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        
        with open(output_file, 'w') as f:
            f.write(response.text)
        
        df = pd.read_csv(output_file, sep='\t', header=None)
        df.columns = ['reactome_id', 'pathway_name', 'species']
        
        print(f"✓ Downloaded pathway list")
        print(f"  Shape: {df.shape}")
        print(f"  Unique species: {df['species'].nunique()}\n")
        
        return df
    
    except Exception as e:
        print(f"✗ Error: {e}")
        return None

def load_layers_from_folder(layers_dir: str) -> dict:
    """
    Load all multiplex layers from a folder of .pkl files.
    
    Parameters
    ----------
    layers_dir : path to folder containing *_layer.pkl files
    
    Returns
    -------
    dict mapping cell_type -> nx.Graph
    """
    layers = {}
    pkl_files = [f for f in os.listdir(layers_dir) if f.endswith('_layer.pkl')]
    
    if not pkl_files:
        raise FileNotFoundError(f"No *_layer.pkl files found in {layers_dir}")
    
    print(f"Found {len(pkl_files)} layer files in {layers_dir}")
    
    for fname in sorted(pkl_files):
        cell_type = fname.replace('_layer.pkl', '').replace('_', ' ')
        fpath = os.path.join(layers_dir, fname)
        with open(fpath, 'rb') as f:
            layers[cell_type] = pickle.load(f)
    
    print(f"✓ Loaded {len(layers)} layers")
    return layers


def get_multiplex_proteins(layers: dict) -> set:
    """Extract all unique proteins (node labels) across all layers."""
    proteins = set()
    for layer in layers.values():
        proteins.update(layer.nodes())
    print(f"✓ {len(proteins):,} unique proteins across all layers")
    return proteins

def merge_interactions_with_pathways(interactions_df, uniprot_pathways_df, multiplex_proteins=None):
    """
    Merge interactions with pathway information.
    Optionally filter to only proteins present in the multiplex.
    """
    print("Merging interactions with pathway information...")

    # ── NEW: subset everything to multiplex proteins first ──────────────────
    if multiplex_proteins is not None:
        print(f"  Filtering to {len(multiplex_proteins):,} multiplex proteins...")
        
        interactions_df = interactions_df[
            interactions_df['# Interactor 1 uniprot id'].str.replace('uniprotkb:', '').isin(multiplex_proteins) &
            interactions_df['Interactor 2 uniprot id'].str.replace('uniprotkb:', '').isin(multiplex_proteins)
        ].copy()
        print(f"  Interactions after filter: {len(interactions_df):,}")

        uniprot_pathways_df = uniprot_pathways_df[
            uniprot_pathways_df['uniprot_id'].isin(multiplex_proteins)
        ].copy()
        print(f"  Pathway mappings after filter: {len(uniprot_pathways_df):,}")
    # ────────────────────────────────────────────────────────────────────────

    interactions_df['uniprot1'] = interactions_df['# Interactor 1 uniprot id'].str.replace('uniprotkb:', '')
    interactions_df['uniprot2'] = interactions_df['Interactor 2 uniprot id'].str.replace('uniprotkb:', '')

    merged = interactions_df.merge(
        uniprot_pathways_df[['uniprot_id', 'pathway_name']],
        left_on='uniprot1', right_on='uniprot_id', how='left'
    ).rename(columns={'pathway_name': 'pathway_protein1'})

    merged = merged.merge(
        uniprot_pathways_df[['uniprot_id', 'pathway_name']],
        left_on='uniprot2', right_on='uniprot_id', how='left'
    ).rename(columns={'pathway_name': 'pathway_protein2'})

    merged['common_pathway'] = merged.apply(
        lambda row: row['pathway_protein1'] if row['pathway_protein1'] == row['pathway_protein2'] else None,
        axis=1
    )

    print(f"✓ Merged {len(merged):,} interactions with pathway information")
    print(f"  Interactions with common pathway: {merged['common_pathway'].notna().sum():,}\n")

    return merged
# ============================================================================
# 4. MERGE INTERACTIONS WITH PATHWAY INFORMATION
# ============================================================================

def merge_interactions_with_pathways(interactions_df, uniprot_pathways_df, multiplex_proteins=None):
    """
    Merge interactions with pathway information.
    Optionally filter to only proteins present in the multiplex.
    """
    print("Merging interactions with pathway information...")

    # ── NEW: subset everything to multiplex proteins first ──────────────────
    if multiplex_proteins is not None:
        print(f"  Filtering to {len(multiplex_proteins):,} multiplex proteins...")
        
        interactions_df = interactions_df[
            interactions_df['# Interactor 1 uniprot id'].str.replace('uniprotkb:', '').isin(multiplex_proteins) &
            interactions_df['Interactor 2 uniprot id'].str.replace('uniprotkb:', '').isin(multiplex_proteins)
        ].copy()
        print(f"  Interactions after filter: {len(interactions_df):,}")

        uniprot_pathways_df = uniprot_pathways_df[
            uniprot_pathways_df['uniprot_id'].isin(multiplex_proteins)
        ].copy()
        print(f"  Pathway mappings after filter: {len(uniprot_pathways_df):,}")
    # ────────────────────────────────────────────────────────────────────────

    interactions_df['uniprot1'] = interactions_df['# Interactor 1 uniprot id'].str.replace('uniprotkb:', '')
    interactions_df['uniprot2'] = interactions_df['Interactor 2 uniprot id'].str.replace('uniprotkb:', '')

    merged = interactions_df.merge(
        uniprot_pathways_df[['uniprot_id', 'pathway_name']],
        left_on='uniprot1', right_on='uniprot_id', how='left'
    ).rename(columns={'pathway_name': 'pathway_protein1'})

    merged = merged.merge(
        uniprot_pathways_df[['uniprot_id', 'pathway_name']],
        left_on='uniprot2', right_on='uniprot_id', how='left'
    ).rename(columns={'pathway_name': 'pathway_protein2'})

    merged['common_pathway'] = merged.apply(
        lambda row: row['pathway_protein1'] if row['pathway_protein1'] == row['pathway_protein2'] else None,
        axis=1
    )

    print(f"✓ Merged {len(merged):,} interactions with pathway information")
    print(f"  Interactions with common pathway: {merged['common_pathway'].notna().sum():,}\n")

    return merged
```

```python
import pandas as pd
import pickle
import os
import networkx as nx

def load_layers_from_folder(layers_dir):
    """Load all *_layer.pkl files from a folder."""
    layers = {}
    pkl_files = [f for f in os.listdir(layers_dir) if f.endswith('_layer.pkl')]
    
    if not pkl_files:
        raise FileNotFoundError(f"No *_layer.pkl files found in {layers_dir}")
    
    print(f"Found {len(pkl_files)} layer files")
    
    for fname in sorted(pkl_files):
        cell_type = fname.replace('_layer.pkl', '')
        fpath = os.path.join(layers_dir, fname)
        with open(fpath, 'rb') as f:
            layers[cell_type] = pickle.load(f)
    
    print(f"✓ Loaded {len(layers)} layers\n")
    return layers


def convert_layer_nodes_to_uniprot(layers, uniprot_to_symbol):
    """
    Convert all node labels in layers from gene symbols to UniProt IDs.
    
    Parameters
    ----------
    layers : dict of cell_type -> nx.Graph (nodes are gene symbols)
    uniprot_to_symbol : dict mapping uniprot_id -> gene_symbol
    
    Returns
    -------
    converted_layers : dict of cell_type -> nx.Graph (nodes are uniprot IDs)
    unmapped_genes : set of genes that couldn't be mapped
    """
    
    # Reverse the mapping: symbol -> uniprot
    symbol_to_uniprot = {v.upper(): k for k, v in uniprot_to_symbol.items()}
    
    converted_layers = {}
    unmapped_genes = set()
    total_mapped = 0
    
    for cell_type, layer in layers.items():
        node_mapping = {}
        for node in layer.nodes():
            symbol_upper = str(node).upper()
            if symbol_upper in symbol_to_uniprot:
                node_mapping[node] = symbol_to_uniprot[symbol_upper]
                total_mapped += 1
            else:
                unmapped_genes.add(node)
                node_mapping[node] = node  # keep original if not found
        
        converted_layer = nx.relabel_nodes(layer, node_mapping)
        converted_layers[cell_type] = converted_layer
        
        n_mapped = sum(1 for v in node_mapping.values() if v in symbol_to_uniprot.values())
        print(f"  {cell_type}: {len(node_mapping)} nodes → {n_mapped} mapped to UniProt")
    
    if unmapped_genes:
        print(f"\n⚠️  {len(unmapped_genes)} genes could not be mapped:")
        for gene in sorted(list(unmapped_genes)[:10]):
            print(f"    - {gene}")
        if len(unmapped_genes) > 10:
            print(f"    ... and {len(unmapped_genes) - 10} more")
    
    return converted_layers, unmapped_genes


def save_converted_layers(converted_layers, output_dir='layers_uniprot'):
    """Save converted layers to new folder."""
    os.makedirs(output_dir, exist_ok=True)
    
    for cell_type, layer in converted_layers.items():
        safe_name = cell_type.replace('/', '_').replace(' ', '_')
        fpath = os.path.join(output_dir, f"{safe_name}_layer.pkl")
        with open(fpath, 'wb') as f:
            pickle.dump(layer, f, protocol=4)
    
    print(f"✓ Saved {len(converted_layers)} converted layers to {output_dir}\n")
    return output_dir


def get_multiplex_proteins(layers):
    """Extract all unique proteins (node labels) across all layers."""
    proteins = set()
    for layer in layers.values():
        proteins.update(layer.nodes())
    print(f"✓ {len(proteins):,} unique proteins in multiplex\n")
    return proteins



```
