```python
# fix_config.py - run this ONCE on the original multixrank_input
import yaml
from pathlib import Path

wdir = Path("multixrank_input")
cfg_path = wdir / "config.yml"

with open(cfg_path) as f:
    cfg = yaml.safe_load(f)

existing = cfg["multiplex"]["gene_multiplex"]["layers"]

# Check which of these actually exist
to_add = []
for stem in ["ppi_layer", "ligand_receptor_layer", "pathway_layer"]:
    tsv = wdir / "multiplex" / "gene_multiplex" / f"{stem}.tsv"
    rel = f"multiplex/gene_multiplex/{stem}.tsv"
    if tsv.exists() and rel not in existing:
        to_add.append(rel)
        print(f"Adding: {rel}")
    elif not tsv.exists():
        print(f"File missing: {tsv}")
    else:
        print(f"Already in config: {rel}")

if to_add:
    new_layers = sorted(existing + to_add)
    n = len(new_layers)
    tau = round(1.0 / n, 6)
    
    cfg["multiplex"]["gene_multiplex"]["layers"] = new_layers
    cfg["multiplex"]["gene_multiplex"]["graph_type"] = ["00"] * n
    cfg["multiplex"]["gene_multiplex"]["tau"] = [tau] * n
    
    with open(cfg_path, "w") as f:
        yaml.safe_dump(cfg, f, default_flow_style=False, sort_keys=False)
    print(f"\nUpdated config: {n} layers, tau={tau}")
else:
    print("Nothing to add")
```
