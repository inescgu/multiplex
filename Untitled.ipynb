```python
#!/usr/bin/env python3
"""
Diagnostic script to check why DRG neurons have 0 edges
"""

import pandas as pd
import pickle
import os

print("="*80)
print("DIAGNOSTIC: Checking DRG Neuron Edge Issue")
print("="*80)

# 1. Check expression file gene IDs
print("\n1. Checking expression file gene IDs...")
expr_file = "../R/processed/drgn_celltype_avg_expression.csv"
if os.path.exists(expr_file):
    expr_df = pd.read_csv(expr_file, index_col=0)
    print(f"   Expression file has {len(expr_df)} genes")
    print(f"   First 5 gene IDs:")
    for i, gene in enumerate(list(expr_df.index[:5])):
        print(f"      {i+1}. '{gene}' (type: {type(gene).__name__})")
    print(f"   Gene ID examples: {list(expr_df.index[:3])}")
else:
    print(f"   ERROR: File not found: {expr_file}")

# 2. Check PPI network gene IDs
print("\n2. Checking PPI network gene IDs...")
ppi_file = "../og_gene_set/temp/combined/full_network.pkl"
if os.path.exists(ppi_file):
    with open(ppi_file, 'rb') as f:
        G = pickle.load(f)
    print(f"   PPI network has {G.number_of_nodes()} nodes")
    print(f"   First 5 gene IDs:")
    for i, gene in enumerate(list(G.nodes())[:5]):
        print(f"      {i+1}. '{gene}' (type: {type(gene).__name__})")
    print(f"   Gene ID examples: {list(G.nodes())[:3]}")
else:
    print(f"   ERROR: File not found: {ppi_file}")

# 3. Check for overlap
print("\n3. Checking overlap...")
if os.path.exists(expr_file) and os.path.exists(ppi_file):
    expr_genes = set(expr_df.index)
    ppi_genes = set(G.nodes())
    
    overlap = expr_genes.intersection(ppi_genes)
    print(f"   Expression genes: {len(expr_genes)}")
    print(f"   PPI genes: {len(ppi_genes)}")
    print(f"   Overlap: {len(overlap)} genes")
    print(f"   Overlap percentage: {len(overlap)/len(expr_genes)*100:.1f}% of expression genes")
    
    if len(overlap) < 100:
        print(f"\n   ⚠️  WARNING: Very low overlap! Only {len(overlap)} genes match!")
        print(f"   This explains why DRG neurons have 0 edges.")
        
        # Show examples of non-matching genes
        print("\n   Sample expression genes NOT in PPI:")
        expr_only = list(expr_genes - ppi_genes)[:5]
        for gene in expr_only:
            print(f"      - '{gene}'")
        
        print("\n   Sample PPI genes NOT in expression:")
        ppi_only = list(ppi_genes - expr_genes)[:5]
        for gene in ppi_only:
            print(f"      - '{gene}'")
    else:
        print(f"\n   ✓ Good overlap! {len(overlap)} genes found in both.")

# 4. Check expression thresholds
print("\n4. Checking expression thresholds...")
if os.path.exists(expr_file):
    print(f"   Cell types in expression file: {list(expr_df.columns)}")
    
    # For first cell type, how many genes pass different thresholds?
    first_celltype = expr_df.columns[0]
    print(f"\n   Checking thresholds for '{first_celltype}':")
    
    for threshold in [0.1, 0.3, 0.5, 0.8, 1.0]:
        n_passing = sum(expr_df[first_celltype] >= threshold)
        print(f"      Threshold {threshold}: {n_passing} genes pass")
    
    # Check if any genes have high expression
    max_expr = expr_df.max().max()
    print(f"\n   Maximum expression value across all cell types: {max_expr:.3f}")
    
    if max_expr < 0.5:
        print(f"   ⚠️  WARNING: Maximum expression is only {max_expr:.3f}")
        print(f"   Default threshold of 0.5 may be too high!")

# 5. Check detection rates if available
print("\n5. Checking detection rates...")
detection_file = "../R/processed/drgn_celltype_detection_rate.csv"
if os.path.exists(detection_file):
    detection_df = pd.read_csv(detection_file, index_col=0)
    print(f"   Detection rate file has {len(detection_df)} genes")
    print(f"   First cell type: {detection_df.columns[0]}")
    
    first_celltype = detection_df.columns[0]
    for threshold in [0.01, 0.05, 0.10, 0.20]:
        n_passing = sum(detection_df[first_celltype] >= threshold)
        print(f"      Detection threshold {threshold}: {n_passing} genes pass")
else:
    print(f"   Detection rate file not found (optional): {detection_file}")

# 6. Suggest fixes
print("\n" + "="*80)
print("DIAGNOSIS COMPLETE")
print("="*80)

if os.path.exists(expr_file) and os.path.exists(ppi_file):
    overlap = len(expr_genes.intersection(ppi_genes))
    max_expr = expr_df.max().max()
    
    print("\nPotential Issues:")
    
    if overlap < 1000:
        print("\n❌ CRITICAL: Gene ID mismatch!")
        print("   - Expression genes and PPI genes use different ID formats")
        print("   - Need to convert gene IDs to match")
        print("\n   Solutions:")
        print("   1. Convert expression gene IDs to match PPI format")
        print("   2. Convert PPI gene IDs to match expression format")
        print("   3. Use a gene ID mapping file (e.g., Ensembl to Symbol)")
    
    if max_expr < 1.0:
        print(f"\n⚠️  WARNING: Low expression values (max={max_expr:.3f})")
        print("   - Default threshold of 0.5 may be too stringent")
        print("\n   Solutions:")
        print(f"   1. Lower expression threshold to {max_expr/2:.2f}")
        print("   2. Check if expression data is log-transformed")
        print("   3. Check normalization method")
    
    if overlap >= 1000 and max_expr >= 0.5:
        print("\n✓ Everything looks good!")
        print("   If still seeing 0 edges, check:")
        print("   1. Are gene IDs exact matches (case-sensitive)?")
        print("   2. Run with --expression-threshold 0.1 to be more lenient")

else:
    print("\n❌ Cannot complete diagnosis - missing required files")

print("\n" + "="*80)
```

```python
#!/usr/bin/env python3
"""
Diagnose why MultiXrank can't find seeds
Run this in your multixrank_setup directory
"""

import os
import sys

def check_genes_in_tsv():
    """Check if problem genes are in the TSV network files"""
    
    problem_seeds = {'FAAH', 'PCSK6', 'LYST', 'CHRM2', 'UCP2', 'TAC1', 'STOML3', 
                     'P2RX3', 'SLC6A1', 'TRPM2', 'PTGDR', 'CALB1', 'EDNRA', 'FRMPD4', 
                     'ANO1', 'SLC6A6', 'NTSR2', 'KLF7', 'CHRNA7', 'ACHE', 'KCNJ3'}
    
    print("="*80)
    print("CHECKING FOR PROBLEM SEEDS IN TSV FILES")
    print("="*80)
    
    # Get all TSV files
    multiplex_dir = 'multiplex/1'
    if not os.path.exists(multiplex_dir):
        print(f"ERROR: Directory not found: {multiplex_dir}")
        print("Make sure you're in the multixrank_setup directory!")
        return False
    
    tsv_files = sorted([f for f in os.listdir(multiplex_dir) if f.endswith('.tsv')])
    print(f"\nFound {len(tsv_files)} TSV files\n")
    
    # Collect all genes across all TSVs
    all_genes = set()
    gene_layer_map = {}  # gene -> [layers it appears in]
    
    for tsv_file in tsv_files:
        path = os.path.join(multiplex_dir, tsv_file)
        genes_in_file = set()
        
        with open(path, 'r') as f:
            for line in f:
                parts = line.strip().split('\t')
                if len(parts) >= 2:
                    source = parts[0]
                    target = parts[1]
                    genes_in_file.add(source)
                    genes_in_file.add(target)
        
        layer_name = tsv_file.replace('.tsv', '')
        for gene in genes_in_file:
            all_genes.add(gene)
            if gene not in gene_layer_map:
                gene_layer_map[gene] = []
            gene_layer_map[gene].append(layer_name)
    
    print(f"Total unique genes across all TSV files: {len(all_genes)}\n")
    
    # Check problem seeds
    print("Checking problem seeds:")
    found = 0
    not_found = 0
    
    for seed in sorted(problem_seeds):
        if seed in all_genes:
            n_layers = len(gene_layer_map[seed])
            print(f"  ✓ {seed:15} - Found in {n_layers} layer(s)")
            found += 1
        else:
            print(f"  ✗ {seed:15} - NOT FOUND in any layer")
            not_found += 1
    
    print(f"\nSummary: {found} found, {not_found} not found")
    
    if not_found > 0:
        print("\n" + "="*80)
        print("THESE GENES ARE NOT IN THE NETWORK")
        print("="*80)
        print("Possible reasons:")
        print("  1. They don't appear in any layer (expression too low)")
        print("  2. They weren't included when TSV files were created")
        print("  3. Gene ID format mismatch")
        print("\nTo include more genes, try:")
        print("  - Lower the --expression-threshold in build_multiplex_interactome.py")
        print("  - Lower the --detection-rate-threshold")
        print("="*80)
    
    return not_found == 0

if __name__ == "__main__":
    success = check_genes_in_tsv()
    sys.exit(0 if success else 1)
```
